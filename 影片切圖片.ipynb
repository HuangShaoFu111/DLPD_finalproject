```python
import cv2
import os

from google.colab import drive
drive.mount('/content/drive')

video_path = "/content/drive/MyDrive/deep learning/期末專題/中文書倒放.MOV 的副本"  #影片位置
save_dir = "/content/drive/MyDrive/deep learning/期末專題/圖片"    #存入資料夾位置
os.makedirs(save_dir, exist_ok=True)

cap = cv2.VideoCapture(video_path)
count = 0

while True:
    ret, frame = cap.read()
    if not ret:
        break
    # 每隔 15 幀存一張（可調整）
    if count % 12 == 0:
        filename = os.path.join(save_dir, f"中文書倒放_{count:04d}.jpg")
        cv2.imwrite(filename, frame)
    count += 1

cap.release()
print("擷取完成 ✅")
```
