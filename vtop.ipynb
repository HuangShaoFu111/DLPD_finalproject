```python
import cv2
import os
import numpy as np

# 使用原始字串 (raw string) 並確保路徑正確
video_path = r"C:\Users\User\Desktop\中央可憐蟲\大四上\深度學習專案設計\DLPD_finalproject\new_V\787663346.594283.mp4"
save_dir = r"C:\Users\User\Desktop\中央可憐蟲\大四上\深度學習專案設計\DLPD_finalproject\picture"

# 檢查影片檔案是否存在
if not os.path.exists(video_path):
    print(f"❌ 錯誤：找不到影片檔案！\n路徑：{video_path}")
else:
    print(f"✅ 找到影片檔案：{video_path}")
    os.makedirs(save_dir, exist_ok=True)

    cap = cv2.VideoCapture(video_path)
    
    if not cap.isOpened():
        print("❌ 錯誤：無法開啟影片檔案 (Codec 可能不支援或檔案損毀)")
    else:
        total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
        print(f"影片開啟成功，總幀數：{total_frames}")
        
        count = 0
        saved_count = 0
        
        while True:
            ret, frame = cap.read()
            if not ret:
                break
            
            # 每隔 12 幀存一張
            if count % 5 == 0:
                filename = os.path.join(save_dir, f"data_{count:04d}.jpg")
                
                # --- 修正：使用 imencode + tofile 解決中文路徑問題 ---
                try:
                    is_success, im_buf = cv2.imencode(".jpg", frame)
                    if is_success:
                        im_buf.tofile(filename)
                        saved_count += 1
                    else:
                        print(f"⚠️ 編碼失敗：{filename}")
                except Exception as e:
                    print(f"⚠️ 寫入例外：{e}")
                # ----------------------------------------------------
            
            count += 1
        
        cap.release()
        print(f"擷取完成 ✅ 共儲存 {saved_count} 張圖片到 {save_dir}")
```
